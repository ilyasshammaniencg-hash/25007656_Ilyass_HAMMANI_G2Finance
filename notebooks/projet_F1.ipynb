```python
# Importations de base
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_openml

# Configuration visuelle
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# ==========================================
# B. Comprendre les données
# ==========================================
# Source : UCI Machine Learning Repository (via OpenML)
# URL : https://archive.ics.uci.edu/ml/datasets/statlog+(german+credit+data)
# Licence : Open Data (Public Domain)

print("Téléchargement des données en cours...")
# Récupération directe sans fichier local
credit = fetch_openml(name='credit-g', version=1, as_frame=True, parser='auto')
df = credit.frame

print(f"Dimensions du dataset : {df.shape[0]} observations et {df.shape[1]} variables.\n")

# Affichage des 5 premières lignes
display(df.head())

# Affichage des informations des variables (types et valeurs manquantes)
df.info()
```
